## LeNet-5

image.png

image.png

Esta `CNN` es alimentada por imágenes de 32x32 píxeles de dígitos manuscritos, y está formada por tres capas convolucionales, dos de las cuales reducen la dimensionalidad de los mapas de características mediante *average pooling*. La salida de la última capa convolucional es conectada un perceptrón multicapa de dos capas, que tienen la responsabilidad de dar la predicción final. [Artículo de publicación original](http://yann.lecun.com/exdb/publis/pdf/lecun-01a.pdf) . A continuación puedes ver un ejemplo de implementación en `Pytorch`.

```python
import torch

def block(c_in, c_out, k=3, p=1, s=1):
    return torch.nn.Sequential(
        torch.nn.Conv2d(c_in, c_out, k, padding=p, stride=s),
        torch.nn.Tanh(),
        torch.nn.AvgPool2d(2, stride=2)
    )

def block2(c_in, c_out):
    return torch.nn.Sequential(
        torch.nn.Linear(c_in, c_out),
        torch.nn.ReLU()
    )

class LeNet5(torch.nn.Module):
  def __init__(self, n_channels=1, n_outputs=10):
    super().__init__()
    #self.pad = torch.nn.ConstantPad2d(2, 0.)
    self.conv1 = block(n_channels, 6, 5, p=0)
    self.conv2 = block(6, 16, 5, p=0)
    self.conv3 = torch.nn.Sequential(
        torch.nn.Conv2d(16, 120, 5, padding=0),
        torch.nn.Tanh()
    )
    self.fc1 = block2(120, 84)
    self.fc2 = torch.nn.Linear(84, 10)

  def forward(self, x):
    #x = self.pad(x)
    x = self.conv1(x)
    x = self.conv2(x)
    x = self.conv3(x)
    x = x.view(x.shape[0], -1)
    x = self.fc1(x)
    x = self.fc2(x)
    return x
```

### 3.1 LeNet-5 (LeCun et al., 1998)

## Actividad
Escribe un parrafo resaltando las diferencias de la arquitectura adaptada vs original

```python
class LeNet5(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 6, kernel_size=5, padding=2),   # 32x32 -> 32x32
            nn.Tanh(),
            nn.AvgPool2d(2),                              # -> 16x16
            nn.Conv2d(6, 16, kernel_size=5),               # -> 12x12
            nn.Tanh(),
            nn.AvgPool2d(2),                               # -> 6x6
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(16 * 6 * 6, 120), nn.Tanh(),
            nn.Linear(120, 84), nn.Tanh(),
            nn.Linear(84, num_classes),
        )

    def forward(self, x):
        x = self.features(x)
        return self.classifier(x)
```

La capa de entrada está diseñada para recibir 3 canales (RGB) en lugar de 1 canal en escala de grises, y se añade un padding=2 en la primera capa convolucional, lo que preserva el tamaño espacial de 32x32 (a diferencia del original, que reducía la dimensión a 28x28). Como consecuencia, las dimensiones espaciales a lo largo de la red cambian: tras el primer pooling se obtiene 16x16, la segunda convolución reduce a 12x12 (en lugar de 10x10) y el segundo pooling resulta en 6x6 (en lugar de 5x5). Esto altera la entrada a la capa Flatten, que ahora aplana 576 características (16 * 6 * 6) en lugar de las 400 originales (16 * 5 * 5). Finalmente, aunque se conserva la función de activación Tanh, la capa de salida utiliza una capa Linear estándar con parámetro num_classes en lugar de la función de base radial (RBF) original, y se omiten las conexiones específicas de la capa C5 original, simplificando la arquitectura para el entrenamiento moderno con retropropagación estándar.

### 3.2 AlexNet adaptado (Krizhevsky et al., 2012)

## Actividad
Escribe un parrafo resaltando las diferencias de la arquitectura adaptada vs original

```python
class AlexNetAdaptado(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=3, padding=1), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),                                        # -> 16x16
            nn.Conv2d(64, 192, kernel_size=3, padding=1), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),                                        # -> 8x8
            nn.Conv2d(192, 384, kernel_size=3, padding=1), nn.ReLU(inplace=True),
            nn.Conv2d(384, 256, kernel_size=3, padding=1), nn.ReLU(inplace=True),
            nn.Conv2d(256, 256, kernel_size=3, padding=1), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),                                        # -> 4x4
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.5),
            nn.Linear(256 * 4 * 4, 1024), nn.ReLU(inplace=True),
            nn.Dropout(0.5),
            nn.Linear(1024, 512), nn.ReLU(inplace=True),
            nn.Linear(512, num_classes),
        )

    def forward(self, x):
        x = self.features(x)
        return self.classifier(x)
```

Esta arquitectura esta diseñada para procesar imágenes de menor resolución (como 32x32) en lugar de las 224x224 originales. En la sección de extracción de características, la adaptación reemplaza el masivo kernel de 11x11 de la primera capa convolucional original por uno mucho más pequeño de 3x3 con padding de 1 y solo 64 filtros (en lugar de 96). Asimismo, todas las capas convolucionales subsecuentes utilizan uniformemente kernels de 3x3 con padding de 1, eliminando por completo las capas de Normalización de Respuesta Local (LRN) que eran distintivas del diseño original. En cuanto al clasificador, la adaptación reduce drásticamente la complejidad: la entrada aplanada es de 4x4 (en lugar de 6x6), y las capas totalmente conectadas se reducen a 1024 y 512 neuronas, muy lejos de las dos capas masivas de 4096 neuronas del AlexNet original. Finalmente, se introduce una capa de Dropout justo después de la operación de Flatten, lo que, junto con la reducción de parámetros, hace que este modelo adaptado sea mucho más ligero y eficiente para conjuntos de datos más pequeños.



### 3.3 VGG16 adaptado (Simonyan & Zisserman, 2014)

## Actividad
Escribe un parrafo resaltando las diferencias de la arquitectura adaptada vs original

```python
class VGGBlock(nn.Module):
    def __init__(self, in_channels, out_channels, num_convs):
        super().__init__()
        layers = []
        for i in range(num_convs):
            in_c = in_channels if i == 0 else out_channels
            layers += [nn.Conv2d(in_c, out_channels, kernel_size=3, padding=1), nn.ReLU(inplace=True)]
        layers.append(nn.MaxPool2d(2))
        self.block = nn.Sequential(*layers)

    def forward(self, x):
        return self.block(x)


class VGG16Adaptado(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.features = nn.Sequential(
            VGGBlock(3, 32, 2),     # 32x32 -> 16x16
            VGGBlock(32, 64, 2),    # -> 8x8
            VGGBlock(64, 128, 3),   # -> 4x4
            VGGBlock(128, 256, 3),  # -> 2x2
            VGGBlock(256, 256, 3),  # -> 1x1
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256 * 1 * 1, 512), nn.ReLU(inplace=True), nn.Dropout(0.5),
            nn.Linear(512, num_classes),
        )

    def forward(self, x):
        x = self.features(x)
        return self.classifier(x)
```

Está diseñada para procesar imágenes de menor resolución (32x32) en lugar de las 224x224 originales. En la sección de extracción de características, aunque se mantiene la profundidad de 13 capas convolucionales organizadas en bloques (2, 2, 3, 3, 3) con kernels de 3x3, el uso de un MaxPool2d(2) tras cada bloque reduce progresivamente la dimensión espacial hasta obtener un mapa de características de 1x1 (en lugar del 7x7 original). La diferencia más drástica radica en el clasificador: la adaptación elimina las tres capas totalmente conectadas masivas del VGG16 original (que tenían 4096, 4096 y 1000 neuronas) y las reemplaza por una arquitectura mucho más ligera que consta de una capa Flatten, una capa oculta de solo 512 neuronas con activación ReLU, una capa de Dropout(0.5) y la capa de salida final. Como resultado, el número total de parámetros se reduce drásticamente, ya que la entrada aplanada es de solo 256 características (256 * 1 * 1) en lugar de las 25,088 originales (512 * 7 * 7), y las capas densas son exponencialmente más pequeñas, haciendo que este modelo adaptado sea computacionalmente mucho más eficiente para conjuntos de datos pequeños.



### 3.4 GoogLeNet mini (Szegedy et al., 2014)

## Actividad
Escribe un parrafo resaltando las diferencias de la arquitectura adaptada vs original

```python
class InceptionBlock(nn.Module):
    def __init__(self, in_channels, ch1x1, ch3x3red, ch3x3, ch5x5red, ch5x5, pool_proj):
        super().__init__()
        self.branch1 = nn.Sequential(
            nn.Conv2d(in_channels, ch1x1, kernel_size=1), nn.ReLU(inplace=True))
        self.branch2 = nn.Sequential(
            nn.Conv2d(in_channels, ch3x3red, kernel_size=1), nn.ReLU(inplace=True),
            nn.Conv2d(ch3x3red, ch3x3, kernel_size=3, padding=1), nn.ReLU(inplace=True))
        self.branch3 = nn.Sequential(
            nn.Conv2d(in_channels, ch5x5red, kernel_size=1), nn.ReLU(inplace=True),
            nn.Conv2d(ch5x5red, ch5x5, kernel_size=5, padding=2), nn.ReLU(inplace=True))
        self.branch4 = nn.Sequential(
            nn.MaxPool2d(kernel_size=3, stride=1, padding=1),
            nn.Conv2d(in_channels, pool_proj, kernel_size=1), nn.ReLU(inplace=True))

    def forward(self, x):
        return torch.cat([self.branch1(x), self.branch2(x), self.branch3(x), self.branch4(x)], dim=1)


class GoogLeNetMini(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.stem = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=3, padding=1), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),                                    # -> 16x16
        )
        self.inception1 = InceptionBlock(64, 32, 32, 64, 8, 16, 16)     # out = 128
        self.inception2 = InceptionBlock(128, 64, 64, 96, 16, 32, 32)   # out = 224
        self.pool = nn.MaxPool2d(2)                                     # -> 8x8
        self.inception3 = InceptionBlock(224, 96, 64, 128, 16, 32, 32)  # out = 288
        self.global_pool = nn.AdaptiveAvgPool2d(1)
        self.classifier = nn.Sequential(nn.Flatten(), nn.Dropout(0.4), nn.Linear(288, num_classes))

    def forward(self, x):
        x = self.stem(x)
        x = self.inception1(x)
        x = self.inception2(x)
        x = self.pool(x)
        x = self.inception3(x)
        x = self.global_pool(x)
        return self.classifier(x)
```

La arquitectura GoogLeNet mini adaptada presenta diferencias drásticas respecto al modelo original de Szegedy et al. (2014), principalmente porque está diseñada para procesar entradas de menor resolución (32x32) y reducir drásticamente la complejidad computacional. En primer lugar, el tallo inicial (stem) se simplifica enormemente: mientras el original utiliza una secuencia compleja de convoluciones (7x7), pooling, normalización LRN y convoluciones adicionales, la versión adaptada emplea una sola capa Conv2d(3, 64, kernel_size=3) seguida de un MaxPool2d(2). En segundo lugar, la red reduce su profundidad y ancho: el original contiene 9 bloques Inception con hasta 1024 canales, mientras que esta versión solo tiene 3 bloques Inception (inception1, inception2, inception3) y sus canales máximos son de 288. Aunque el InceptionBlock adaptado conserva fielmente las 4 ramas paralelas (1x1, 1x1→3x3, 1x1→5x5 y MaxPool→1x1) y la concatenación, se eliminan por completo los clasificadores auxiliares que usaba el original para combatir el desvanecimiento del gradiente. Finalmente, la cabeza clasificadora cambia: en lugar de las capas totalmente conectadas masivas, se utiliza un AdaptiveAvgPool2d(1) (Global Average Pooling), seguido de un Dropout(0.4) y una única capa Linear(288, num_classes), lo que reduce el número de parámetros de manera exponencial, haciendo el modelo mucho más ligero y adecuado para conjuntos de datos pequeños.

## Actividad
Define hiperparámetros de entrenamiento y optimizador

```python
def count_params(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)


def train_model(model, train_loader, val_loader, epochs=8, lr=1e-3, name="modelo"):
    model = model.to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    history = {"train_loss": [], "train_acc": [], "val_loss": [], "val_acc": []}

    start = time.time()
    for epoch in range(epochs):
        model.train()
        running_loss, correct, total = 0.0, 0, 0
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad()
            out = model(x)
            loss = criterion(out, y)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * x.size(0)
            correct += (out.argmax(1) == y).sum().item()
            total += x.size(0)
        train_loss, train_acc = running_loss / total, correct / total

        model.eval()
        v_loss, v_correct, v_total = 0.0, 0, 0
        with torch.no_grad():
            for x, y in val_loader:
                x, y = x.to(device), y.to(device)
                out = model(x)
                loss = criterion(out, y)
                v_loss += loss.item() * x.size(0)
                v_correct += (out.argmax(1) == y).sum().item()
                v_total += x.size(0)
        val_loss, val_acc = v_loss / v_total, v_correct / v_total

        history["train_loss"].append(train_loss)
        history["train_acc"].append(train_acc)
        history["val_loss"].append(val_loss)
        history["val_acc"].append(val_acc)
        print(f"[{name}] Época {epoch+1}/{epochs} - "
              f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} "
              f"val_loss={val_loss:.4f} val_acc={val_acc:.4f}")

    elapsed = time.time() - start
    print(f"[{name}] Tiempo total de entrenamiento: {elapsed:.1f} s")
    return history, elapsed


def evaluate_test(model, test_loader):
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for x, y in test_loader:
            x, y = x.to(device), y.to(device)
            out = model(x)
            correct += (out.argmax(1) == y).sum().item()
            total += x.size(0)
    return correct / total
```

## 5. Entrenamiento de las cuatro arquitecturas

`EPOCHS` Una buena práctica es configurar un bajo numero. Si el tiempo lo permite, se puede subir a 15–20 épocas para ver mejor las diferencias de convergencia.

```python
EPOCHS = 5

modelos = {
    "LeNet-5": LeNet5(),
    "AlexNet (adaptado)": AlexNetAdaptado(),
    "GoogLeNet (mini)": GoogLeNetMini(),
    "VGG16 (adaptado)": VGG16Adaptado(),
}

historiales = {}
resultados = {}

for nombre, modelo in modelos.items():
    print(f"\n=== Entrenando {nombre} ===")
    n_params = count_params(modelo)
    print(f"Parámetros entrenables: {n_params:,}")
    hist, tiempo = train_model(modelo, train_loader, val_loader, epochs=EPOCHS, name=nombre)
    test_acc = evaluate_test(modelo, test_loader)
    print(f"[{nombre}] Accuracy en test: {test_acc:.4f}")

    historiales[nombre] = hist
    resultados[nombre] = {
        "parametros": n_params,
        "tiempo_entrenamiento_s": round(tiempo, 1),
        "val_acc_final": round(hist["val_acc"][-1], 4),
        "test_accuracy": round(test_acc, 4),
    }
```
